```python
import pandas as pd
```

```python
import os
from pathlib import Path
# Qwen2.5-1.5B-Instruct_countdown2345_grpo_gaussian_0.5_0.5_True_1600
test_set = [f'countdown_n{i}t100_1-100' for i in range(2, 7)]
pass_at_k_file = Path(os.getcwd()).parents[1] / 'outputs' / 'Qwen2.5-1.5B-Instruct_countdown2345_grpo_gaussian_0.25_0.75_True_1200' / f"pass_at_k_results_{test_set[-1]}.csv"
```

```python

df_pass = pd.read_csv(pass_at_k_file).transpose()
df_pass
```

```python
df_pass_at_k = df_pass.cummax()
df_pass_at_k
```

```python
df_mean_pass_at_k = df_pass_at_k.mean(axis=1)
df_mean_pass_at_k
```

```python
df_mean_pass_at_k.index.name = "k"
df_mean_pass_at_k.name = "pass_at_k"
df_mean_pass_at_k
```

```python

df_mean_pass_at_k.plot(y="pass_at_k", ylabel="Pass@k", title="Avg Pass@k")
```
